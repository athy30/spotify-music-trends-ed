## Popularity Distribution

### Track Popularity


### Histogram

```python
plt.figure(figsize = (8,5))
sns.histplot(df['track_popularity'], bins = 20, kde = True)
plt.title("Distribution of Track Popularity")
plt.xlabel("Track popularity")
plt.ylabel("Count")
plt.show()
```

### Boxplot

```python
plt.figure(figsize= (8,4))
sns.boxplot(x = df['track_popularity'])
plt.title("Track Popularity Boxplot")
plt.show()
```

### Artist Popularity

```python
plt.figure(figsize = (8,5))
sns.histplot(df['artist_popularity'], bins = 20, kde = True)
plt.title("Distribution of Artist Popularity")
plt.xlabel("Artist popularity")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize= (8,4))
sns.boxplot(x = df['artist_popularity'])
plt.title("Track Popularity Boxplot")
plt.show()
```

### Track Duration in mins


```python
plt.figure(figsize = (8,5))
sns.histplot(df['track_duration_min'], bins = 20, kde = True)
plt.title("Distribution of Track Duration")
plt.xlabel("Track Duration")
plt.ylabel("Count")
plt.show()
```

### Artist Followers

```python
plt.figure(figsize = (8,5))
sns.histplot(df['artist_followers'], bins = 20)
plt.title("Distribution of Artist followers")
plt.xlabel("Artist Followers")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize = (8,5))
sns.histplot(np.log1p(df['artist_followers']),bins = 30)
plt.title("Log transformed artist followers")
plt.xlabel("Log(Artist Followers)")
plt.ylabel("Count")           
plt.show()
```

### Categorical Column

```python
plt.figure(figsize = (6,4))
sns.countplot(x = 'explicit' , data = df)
plt.title("Explicit vs Non Explicit Songs")
plt.show()
```

### Album type

```python
plt.figure(figsize = (6,4))
sns.countplot(x = 'album_type' , data = df)
plt.title("Album type distribution")
plt.show()
```

### CORRELATION matrix

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only = True), annot = True, cmap = 'coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

### Scatter Plots

#### Artist popularity vs Track popularity

```python
plt.figure(figsize = (8,5))
sns.scatterplot(x = 'artist_popularity' , y= 'track_popularity' , data = df)
plt.title("Artist popularity vs Track popularity")
plt.show()
```

#### Artist followers vs Track Popularity 

```python
plt.figure(figsize = (8,5))
sns.scatterplot(x = np.log1p(df['artist_followers']),
                y = df['track_popularity'])
plt.title("Artist Followers vs Track Popularity (Log Scale)")
plt.xlabel("Log(Artist Followers)")
plt.show()
```

#### Track Duration vs Popularity

```python
plt.figure(figsize = (8,5))
sns.scatterplot(x = 'track_duration_min' , y = 'track_popularity' , data = df)
plt.title("Track Duration vs Popularity")
plt.show()
```

#### Explicit vs Track Popularity

```python
plt.figure(figsize = (6,5))
sns.boxplot(x = 'explicit', y = 'track_popularity', data = df)
plt.title("Explicit vs Track Popularity")
plt.xlabel("Explicit")
plt.ylabel("Track Popularity")
plt.show()
```

#### Album type vs Track Popularity

```python
plt.figure (figsize = (8,5))
sns.boxplot(x = 'album_type' , y = 'track_popularity' , data = df)
plt.title("Album Type vs Track popularity")
plt.xlabel("Album Type")
plt.ylabel("Track Popularity")
plt.show()
```

### Group Analysis

#### Average Popularity by Album Type

```python
df.groupby('album_type')['track_popularity'].mean()
```

#### Average Popularity by Explicit

```python
df.groupby('explicit')['track_popularity'].mean()
```

### Date Analysis

```python
df['album_release_date'] = pd.to_datetime(df['album_release_date'])
df['release_year'] = df['album_release_date'].dt.year
```

#### Number of releases per year

```python
plt.figure(figsize = (10,5))
sns.countplot(x = 'release_year', data = df)
plt.xticks(rotation = 45)
plt.title ("Number of Releases per Year")
plt.show()
```

#### Average Popularity by Year

```python
df.groupby('release_year')['track_popularity'].mean().plot(figsize = (10,5))
plt.title("Average Track Popularity by Year")
plt.ylabel("Average Popularity")
plt.show()
```
